## Step 4 - Clean the data

```python
# 4.1 Remove the extra column that has no name

df = df.drop('Unnamed: 0', axis=1)
print("Columns after removing the extra column:")
print(df.columns.tolist())
```

```python
# 4.2 Remove duplicate rows

print("Rows before removing duplicates:", df.shape[0])

df = df.drop_duplicates()
df = df.reset_index(drop=True)   # reset the row numbers after deleting rows

print("Rows after removing duplicates :", df.shape[0])
```

```python
# 4.3 Remove extra spaces from the text columns

df['Product Name'] = df['Product Name'].str.strip()
df['Brand'] = df['Brand'].str.strip()
df.columns = df.columns.str.strip()

print("Extra spaces removed!")
print(df['Product Name'].head())
```

```python
# 4.4 Fix the brand names so that one brand has only one spelling

df['Brand'] = df['Brand'].replace({
    'MOTOROLA': 'Motorola',
    'REDMI'   : 'Redmi',
    'NOTE'    : 'Redmi',      # "NOTE 12 5G" is actually a Redmi Note 12 5G
    'realme'  : 'Realme',
    'POCO'    : 'Poco',
    'IQOO'    : 'iQOO',
    'OPPO'    : 'Oppo',
    'LAVA'    : 'Lava',
    'itel'    : 'Itel',
    'vivo'    : 'Vivo'
})

print("Total number of brands:", df['Brand'].nunique())
print()
print(df['Brand'].value_counts())
```

```python
# 4.5 Fix the wrong screen size
# A mobile phone cannot have a 17 inch screen, so this value is a mistake.
# We replace it with the middle (median) screen size.

print("Phones with a wrong screen size:")
print(df[df['Screen_size'] > 8][['Product Name', 'Brand', 'Screen_size']])

df.loc[df['Screen_size'] > 8, 'Screen_size'] = df['Screen_size'].median()

print()
print("Smallest screen size now:", df['Screen_size'].min())
print("Largest screen size now :", df['Screen_size'].max())
```

```python
# 4.6 Check the cleaned data one last time

print("Shape of cleaned data:", df.shape)
print("Total missing values :", df.isnull().sum().sum())
print("Total duplicate rows :", df.duplicated().sum())

df.head()
```

## Step 5 - Analyse the data

Now the data is clean. Let us find the answers to the questions we wrote at the start.

### Q1. Which brand has the highest number of phones?

```python
brand_count = df['Brand'].value_counts()

print("Number of phones of each brand:")
print(brand_count)

print()
print("The brand with the most phones is", brand_count.idxmax(),
      "with", brand_count.max(), "phones.")
```

### Q2. What is the price range of the phones?

```python
print("Cheapest phone price :", df['Price'].min(), "rupees")
print("Costliest phone price:", df['Price'].max(), "rupees")
print("Average price        :", round(df['Price'].mean(), 0), "rupees")
print("Middle (median) price:", df['Price'].median(), "rupees")
```

### Q3. Which are the 5 most expensive phones?

```python
df.sort_values('Price', ascending=False).head(5)[['Product Name', 'Brand', 'Price', 'Rating']]
```

### Q4. Which are the 5 cheapest phones?

```python
df.sort_values('Price').head(5)[['Product Name', 'Brand', 'Price', 'Rating']]
```

### Q5. Which brand is the most expensive on an average?

```python
avg_price = df.groupby('Brand')['Price'].mean().round(0).sort_values(ascending=False)

print("Average price of each brand:")
print(avg_price)
```

### Q6. Which brand has the best average rating?

```python
avg_rating = df.groupby('Brand')['Rating'].mean().round(2).sort_values(ascending=False)

print("Average rating of each brand:")
print(avg_rating)
```

### Q7. How many phones are 4G and how many are 5G?

```python
# If the product name has "5G" in it, then it is a 5G phone
df['Network'] = df['Product Name'].apply(lambda name: '5G' if '5G' in name else '4G')

print(df['Network'].value_counts())
```

### Q8. What is the average rating given by customers?

```python
print("Average customer rating :", round(df['Rating'].mean(), 2), "out of 5")
print("Highest rating          :", df['Rating'].max())
print("Lowest rating           :", df['Rating'].min())

print()
print("How many ratings fall in each category:")
print(df['Rating_Category'].value_counts())
```

### Q9. Which phone has the biggest battery?

```python
print("Top 5 phones with the biggest battery:")
df.sort_values('Battery', ascending=False).head(5)[['Product Name', 'Brand', 'Battery', 'Price']]
```

## Step 6 - Charts

A chart makes the numbers much easier to understand than a table.

### Chart 1 - Number of phones of each brand

```python
top_brands = df['Brand'].value_counts().head(10)

plt.figure(figsize=(10, 5))
plt.bar(top_brands.index, top_brands.values, color='steelblue')
plt.title('Top 10 Brands by Number of Phones')
plt.xlabel('Brand')
plt.ylabel('Number of phones')
plt.xticks(rotation=45)
plt.show()
```

### Chart 2 - How are the prices spread?

```python
plt.figure(figsize=(10, 5))
plt.hist(df['Price'], bins=30, color='orange', edgecolor='black')
plt.title('Price Distribution of Mobile Phones')
plt.xlabel('Price in rupees')
plt.ylabel('Number of phones')
plt.show()
```

From this chart we can see that **most phones are cheap** and only a few phones are very
expensive. This kind of chart shape is called **right skewed**.

### Chart 3 - 4G vs 5G phones

```python
network_count = df['Network'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(network_count.values, labels=network_count.index,
        autopct='%1.1f%%', startangle=90, colors=['skyblue', 'lightgreen'])
plt.title('4G vs 5G Phones')
plt.show()
```

### Chart 4 - Average price of the top 10 brands

```python
top10 = df['Brand'].value_counts().head(10).index
avg_price_top10 = df[df['Brand'].isin(top10)].groupby('Brand')['Price'].mean().sort_values()

plt.figure(figsize=(10, 5))
plt.barh(avg_price_top10.index, avg_price_top10.values, color='mediumseagreen')
plt.title('Average Price of Top 10 Brands')
plt.xlabel('Average price in rupees')
plt.ylabel('Brand')
plt.show()
```

### Chart 5 - Does a costlier phone get a better rating?

```python
plt.figure(figsize=(10, 5))
plt.scatter(df['Price'], df['Rating'], color='purple', alpha=0.5)
plt.title('Price vs Rating')
plt.xlabel('Price in rupees')
plt.ylabel('Rating out of 5')
plt.show()

print("Correlation between Price and Rating:", round(df['Price'].corr(df['Rating']), 2))
```

### Chart 6 - Average rating of the top 10 brands

```python
avg_rating_top10 = df[df['Brand'].isin(top10)].groupby('Brand')['Rating'].mean().sort_values()

plt.figure(figsize=(10, 5))
plt.barh(avg_rating_top10.index, avg_rating_top10.values, color='tomato')
plt.title('Average Rating of Top 10 Brands')
plt.xlabel('Average rating out of 5')
plt.ylabel('Brand')
plt.xlim(3.8, 4.8)
plt.show()
```
